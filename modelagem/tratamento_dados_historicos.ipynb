# Carregando a base de dados

```python
import pandas as pd
```

```python
ibovespa = pd.read_csv(r"../data/raw/Dados Históricos - Ibovespa.csv", sep=',')
ibovespa.head()
```

```python
# Quais os tipos informações compõe o banco de dados (base original)
ibovespa.info()
```

# EDA - Análise Exploratória de Dados IBOVESPA

```python
# renomeando as colunas
ibovespa.rename(columns={'Data': 'data', 'Abertura': 'abertura', 'Máxima': 'maxima', 'Mínima': 'minima', 'Último': 'fechamento', 'Vol.': 'volume', "Var%": "variacao_percentual"}, inplace=True)
ibovespa.head()
```

```python
# Transformando a coluna Data de objeto para data
ibovespa["data"] = pd.to_datetime(ibovespa["data"])
print(ibovespa.head())
```

```python
# Ajustando a coluna de variação
ibovespa["variacao_percentual"] = ibovespa["variacao_percentual"].str.replace('%', '').str.replace(',', '.').astype(float)
print(ibovespa.head())
```

```python
# Quantidade de dados nulos por coluna
ibovespa.isnull().sum()
```

```python
# # Criando uma coluna se fechou em alta ou em baixa
# ibovespa['status'] = ibovespa.apply(lambda row: 'alta' if row['variacao_dia_posterior'] > 0 else 'baixa', axis=1)
# ibovespa.head()
```

```python
def convert_volume(volume_str):
    if pd.isna(volume_str):
        return None
    
    volume_str = str(volume_str).strip()
    volume_str = volume_str.replace('.', '').replace(',', '.')
    
    if volume_str.endswith('M'):
        return float(volume_str[:-1]) * 1_000_000
    elif volume_str.endswith('B'):
        return float(volume_str[:-1]) * 1_000_000_000
    else:
        return float(volume_str)

ibovespa['vol_reais'] = ibovespa['volume'].apply(convert_volume)
```

```python
# Validando a saída final da base
ibovespa.info()
```

# Enriquecendo com o dolar do dia

```python
dolar = pd.read_csv(r"../data/raw/USD_BRL Dados Históricos.csv", sep=',')
dolar = dolar[['Data', 'Último']]
dolar.rename(columns={'Data': 'data', 'Último': 'cotacao_dia_dolar'}, inplace=True)
dolar['data'] = pd.to_datetime(dolar['data'])
dolar['cotacao_dia_dolar'] = dolar['cotacao_dia_dolar'].str.replace(',', '.').astype(float)
dolar.head()
```

```python
dolar.info()
```

```python
ibovespa_enriquecido = ibovespa.merge(dolar, on='data', how='left')
```

```python
ibovespa_enriquecido.isnull().sum()
```

# Enriquecendo S&P 500

```python
sep_500 = pd.read_csv(r"../data/raw/Dados Históricos - S&P 500.csv", sep=',')
sep_500 = sep_500[['Data', 'Último']]
sep_500.rename(columns={'Data': 'data', 'Último': 'cotacao_dia_sep_500'}, inplace=True)
sep_500['data'] = pd.to_datetime(sep_500['data'])
sep_500['cotacao_dia_sep_500'] = sep_500['cotacao_dia_sep_500'].str.replace('.', '').str.replace(',', '.').astype(float)
sep_500.head()
```

```python
ibovespa_enriquecido_2 = ibovespa_enriquecido.merge(sep_500, on='data', how='left')
```

# Enriquecendo preço Petróleo

```python
petroleo = pd.read_csv(r"../data/raw/Dados Históricos - Petróleo Brent Futuros.csv", sep=',')
petroleo = petroleo[['Data', 'Último']]
petroleo.rename(columns={'Data': 'data', 'Último': 'cotacao_petroleo'}, inplace=True)
petroleo['data'] = pd.to_datetime(petroleo['data'])
petroleo['cotacao_petroleo'] = petroleo['cotacao_petroleo'].str.replace('.', '').str.replace(',', '.').astype(float)
petroleo.head()
```

```python
ibovespa_enriquecido_3 = ibovespa_enriquecido_2.merge(petroleo, on='data', how='left')
```

# Formatando base final

```python
# Considerando apenas as colunas relevantes
ibovespa_final = ibovespa_enriquecido_3[['data', 'abertura', 'maxima', 'minima', 'fechamento', 'vol_reais', 'variacao_percentual', 'cotacao_dia_dolar', 'cotacao_dia_sep_500', 'cotacao_petroleo']]
```

```python
# # Salvando dataframe final em csv
ibovespa_final.to_csv('../data/trusted/ibovespa_historico.csv', index=False, sep = ';')
```
